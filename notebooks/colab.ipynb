# DINOv2 + RAE a 512×512

No VS Code, liga o notebook a um runtime GPU pela extensão oficial Google Colab.

```python
import subprocess
import sys
from pathlib import Path

subprocess.run(["nvidia-smi"], check=True)
```

```python
repo_url = "https://github.com/andreferraria453/Synthetic-Data-Generation-Framework.git"
repo = Path("/content/rae-experiment")

if not repo.exists():
    subprocess.run(["git", "clone", "--recurse-submodules", repo_url, str(repo)], check=True)
else:
    subprocess.run(["git", "pull", "--ff-only"], cwd=repo, check=True)
subprocess.run(["git", "submodule", "update", "--init", "--recursive"], cwd=repo, check=True)
```

```python
subprocess.run([sys.executable, "-m", "pip", "install", "-r", "requirements.txt"], cwd=repo, check=True)
```

```python
save_to_drive = False
output = repo / "outputs"

if save_to_drive:
    from google.colab import drive

    drive.mount("/content/drive")
    output = Path("/content/drive/MyDrive/rae-experiment")
```

```python
subprocess.run([
    sys.executable, "-m", "rae_experiment.sampling.reconstruct",
    "--output", str(output / "reconstruction.png"),
], cwd=repo, check=True)
```

```python
subprocess.run([
    sys.executable, "-m", "rae_experiment.sampling.generate",
    "--class-id", "207", "--seed", "0", "--output", str(output / "generated.png"),
], cwd=repo, check=True)
```

```python
subprocess.run([
    sys.executable, "-m", "rae_experiment.sampling.generate_sana",
    "--batches", "5", "--csv", str(output / "generation_performance.csv"),
], cwd=repo, check=True)
```

```python
sprint_precision = "fp16"  # T4 smoke test; select bf16 for the A6000.
result = subprocess.run([
    sys.executable, "-m", "rae_experiment.sampling.generate_sana_sprint",
    "--precision", sprint_precision,
    "--batches", "1", "--csv", str(output / "generation_performance.csv"),
], cwd=repo, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
print(result.stdout)
result.check_returncode()
```

```python
subprocess.run([
    sys.executable, "-m", "rae_experiment.sampling.generate_bar",
    "--batches", "5", "--csv", str(output / "generation_performance.csv"),
], cwd=repo, check=True)
```
